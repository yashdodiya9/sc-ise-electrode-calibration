```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("06-17-2026_OCP_CaCl2 calibration_R75-129_75,77,78,80,81,83,92,98,100,102,105,114,117,120,121,122,124,125,126,127,128,129.csv",
                  encoding="utf-16", skiprows=5)
df.head()
```

```python
channels = [i for i in range(1,23)]
electrode = [75,77,78,80,81,83,92,98,100,102,105,114,117,120,121,122,124,125,126,127,128,129]
time_index = 0
potential_indices = []
for i in range(1,df.shape[1],2):
    potential_indices.append(i)
print(len(potential_indices), len(electrode))
```

```python
def clean_naming(df, electrode, time_index, potential_indices, name, t_sec_to_hour=True):
    cols = list(df.columns)
    for i, el in zip(potential_indices, electrode):
        cols[i] = f'R{el}'
    
    print(t_sec_to_hour, type(t_sec_to_hour))
    if t_sec_to_hour == True:
        cols[time_index] = 't_h'
        df.columns = cols
        df['t_h'] = pd.to_numeric(df['t_h'], errors='coerce')
        df['t_h'] = df['t_h'] / 3600
    else:
        cols[time_index] = 't_s'
        df.columns = cols
        df['t_s'] = pd.to_numeric(df['t_s'], errors='coerce')
    df.to_csv(name, index=False)
```

```python
clean_naming(df, electrode=electrode, time_index=time_index, potential_indices=potential_indices, name='clean_name_dataset/06-17-2026_OCP_Calibration_CaCl2', t_sec_to_hour=False)
```
